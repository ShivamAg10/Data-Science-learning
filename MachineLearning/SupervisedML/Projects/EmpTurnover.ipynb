```python
import pandas as pd 
from sklearn.linear_model import LogisticRegression, LinearRegression, Lasso
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score
```

```python
emp = pd.read_csv("D:\desktop\Data Science Learning\Data-Science-learning\MachineLearning\DataSet\employee_turnover.csv")
emp
```

```python
## Checking Null Points
emp.info()
emp.describe()
```

```python
# Checking output column
emp["Employee_Turnover"].unique()
```

```python
# Defining Input Output Column
X = emp.drop(columns=["Employee_Turnover"], axis=1)
Y = emp["Employee_Turnover"]
X.head()
Y.head()
```

```python
## Splitting data
X_Train, X_Test, Y_Train, Y_Test = train_test_split(
    X,
    Y,
    random_state=42,
    test_size=0.2
)
Y_Train
```

```python
## Checking Ouput Values split in y_train
Y_Train[Y_Train == 1] # 547
Y_Train[Y_Train == 0] # 533
```

```python
# Defining & Training Model
model = LogisticRegression()
model.fit(
    X_Train, Y_Train
)
```

```python
## Lasso Model
lasso = LogisticRegression(penalty="l1", solver="liblinear", C=0.5)
lasso.fit(
    X_Train, Y_Train
)
```

```python
## Ridge Model
ridge = LogisticRegression(penalty="l2", C = 1.0, max_iter=200)
ridge.fit(
    X_Train, Y_Train
)
```

```python
# Predicting Values
Y_pred_model = model.predict(X_Test)
Y_pred_lasso = lasso.predict(X_Test)
Y_pred_ridge = ridge.predict(X_Test)
```

```python
## Accuracy and Precision Score
print("Model Accuracy: ", accuracy_score(Y_Test, Y_pred_model))
print("Lasso Accuracy: ", accuracy_score(Y_Test, Y_pred_lasso))
print("Ridge Accuracy: ", accuracy_score(Y_Test, Y_pred_ridge))
```
