```python
import numpy as np 
```

```python
# Linear Regression

class LinearRegression:
    def __init__(self, learning_rate=0.01, n_iter=1000):
        self.bias = None
        self.weights = None
        self.lr = learning_rate
        self.n_iter = n_iter
    
    def fit(self, x, y): # x_train, y_train
        m,n = x.shape ## (no of samples, no of features)
        
        ## Step -1 -> initialization
        self.bias = 0
        self.weights = np.zeros(n)
        
        # Gradient Descent
        for i in range(self.n_iter):
            ## Step -2 -> calc y_pred
            y_pred = self.bias + np.dot(x, self.weights)
            
            ## Step-3 -> calc gradients
            diff_bias = (1/m) * np.sum(y_pred - y)
            diff_weights = (1/m) * np.dot(x.T , (y_pred - y))
            
            ## Step-4 -> Applying Convergence Theorem -> params update
            self.bias -= self.lr*diff_bias
            self.weights -= self.lr*diff_weights
    
    def predict(self, x):
        y_pred = self.bias + np.dot(x, self.weights)
        return y_pred
```

```python
x = np.array([[1], [2], [3], [4], [5]])
y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()
model.fit(x,y)

y_pred = model.predict(x)
print(y_pred)

print(model.bias)
print(model.weights)
```
