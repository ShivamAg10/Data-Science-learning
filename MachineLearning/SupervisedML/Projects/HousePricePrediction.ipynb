```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
```

```python
homeVistaprop = pd.read_csv("D:\desktop\Data Science Learning\Data-Science-learning\MachineLearning\DataSet\HousePricePrediction.csv")
homeVistaprop.head()
homeVistaprop.loc[2100]
```

```python
homeVistaprop.isnull().sum()
```

```python
## Handling Missing Value

homeVistaprop[["MSZoning","Exterior1st"]] = homeVistaprop[["MSZoning","Exterior1st"]].ffill()
homeVistaprop.isnull().sum()

homeVistaprop[homeVistaprop['BsmtFinSF2'].isnull()]
homeVistaprop['BsmtFinSF2'] = homeVistaprop['BsmtFinSF2'].fillna(homeVistaprop['BsmtFinSF2'].median())
homeVistaprop.isnull().sum()

homeVistaprop[homeVistaprop['TotalBsmtSF'].isnull()]
homeVistaprop['TotalBsmtSF'] = homeVistaprop['TotalBsmtSF'].fillna(homeVistaprop['TotalBsmtSF'].median())
homeVistaprop.isnull().sum()

homeVistaprop[homeVistaprop["SalePrice"].isnull()]
homeVistaprop.dropna(subset=['SalePrice'], inplace=True)
```

```python
# homeVistaprop.drop(["Id"], axis=1, inplace=True)
homeVistaprop.head()
```

```python
X = homeVistaprop.drop(["MSZoning", "LotConfig", "BldgType", "Exterior1st", "SalePrice"], axis=1)
Y = homeVistaprop["SalePrice"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=10
)
print(x_train.shape)
```

```python
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
y_pred = model.predict(x_test)
print(y_pred[0])
print(y_test.iloc[0])
```

```python
r2 = r2_score(y_test, y_pred)
r2
```
